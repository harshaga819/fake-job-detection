### 1. Problem Statement

Online job postings are widely used by job seekers and employers. However, many fraudulent postings exist, which can lead to scams, data theft, and financial losses. The goal of this project is to build a machine learning model that can detect whether a job posting is real or fake, based on its textual and categorical attributes.

### 6. Feature Engineering

```python
df_feat = df.copy()
```

#### A. Handling Missing Data

##### Drop columns with more than 50% missing values

```python
threshold = 0.5
```

```python
df_feat = df_feat.loc[:, df_feat.isnull().mean() < threshold]
```

##### For remaining columns:

If categorical → fill with space (' ')


If numeric → fill with 0

```python
for col in df_feat.columns:
    if df_feat[col].dtype == 'object':
        df_feat[col] = df_feat[col].fillna(' ')  
    else:
        df_feat[col] = df_feat[col].fillna(0)    
```

#### B. Feature Extraction (from text)

```python
df_feat['description_length'] = df_feat['description'].str.len()
df_feat['caps_ratio_desc'] = df_feat['description'].apply(lambda x: sum(1 for c in str(x) if c.isupper()) / len(str(x)) if len(str(x)) > 0 else 0)
```

```python
df_feat.head()
```

## Other Apporach

```python
def feature_subset_modeling(feature_cols, df=df_feat, target='fraudulent', tfidf_max_features=5000):
    
    df_model = df[feature_cols + [target]].copy()
    df_model.fillna(' ', inplace=True)
    
    
    df_model['combined_text'] = df_model[feature_cols].apply(lambda row: ' '.join(row.values.astype(str)), axis=1)
    
    X = df_model['combined_text']
    y = df_model[target]
    

    tfidf = TfidfVectorizer(stop_words='english', max_features=tfidf_max_features, ngram_range=(1,2))
    X_tfidf = tfidf.fit_transform(X)

    
    X_train, X_test, y_train, y_test = train_test_split(X_tfidf, y, test_size=0.2, random_state=42, stratify=y)

    
    models = {
        "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42, class_weight='balanced', n_jobs=-1),
        "Linear SVM": LinearSVC(max_iter=2000, random_state=42, class_weight='balanced', dual=False)
    }
    
    print(f"\n===== Running on feature subset: {feature_cols} =====\n")
    
    for name, model in models.items():
        print(f"\n--- {name} ---")
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        
        accuracy = accuracy_score(y_test, y_pred)
        print(f"Accuracy: {accuracy * 100:.2f}%")
        
        print("Classification Report:")
        print(classification_report(y_test, y_pred, target_names=['Real Job (0)', 'Fake Job (1)']))
        
        print("Confusion Matrix:")
        print(confusion_matrix(y_test, y_pred))
```
